Vehicle Service EDA

```python
import pandas as pd
```

```python
#LOAD THE DATASET
df = pd.read_csv("Q48_vehicle_service.csv")
```

```python
#PRINT FIRST 5 RECORDS
print("First 5 Records:")
print(df.head())
```

```python
#DATA SET INFORMATION
print("\nDataset Information:")
print(df.info())
```

```python
print("\nStatistical Summary:")
print(df.describe())
```

```python
#MISSING VALUES
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
#resolving mileage missing values by replacing them with the median value of the column
df['mileage_km'] = df['mileage_km'].fillna(df['mileage_km'].median())
```

```python
#removing duplicate records
print("\nNumber of Duplicate Records:",
      df.duplicated().sum())

df = df.drop_duplicates()
```

```python
#STANDARDIZING THE SERVICE TYPE COLUMN
df["service_type"] = (
    df["service_type"]
    .str.strip()
    .str.lower()
    .str.title()
)
print("\nStandardized Service Types:")
print(df["service_type"].unique())
```

```python
#STANDARDIZING THE REPEAT VISIT COLUMN
df["repeat_visit"] = (
    df["repeat_visit"]
    .astype(str)
    .str.strip()
    .str.lower()
)
df["repeat_visit"] = df["repeat_visit"].replace({
    "yes": "Yes",
    "y": "Yes",
    "true": "Yes",
    "no": "No",
    "n": "No",
    "false": "No"
})
```

```python
def age_category(age):
    if age <= 3:
        return "New"
    elif age <= 8:
        return "Medium"
    else:
        return "Old"

df["vehicle_age_category"] = (
    df["vehicle_age_years"].apply(age_category)
)
```

```python

print("\nService Duration by Vehicle Age:")
print(
    df.groupby("vehicle_age_category")
    ["service_duration_hours"].mean()
)
```

```python
#REPEAT VISITS BY SERVICE TYPE
print("\nRepeat Visits by Service Type:")
print(
    pd.crosstab(
        df["service_type"],
        df["repeat_visit"]
    )
)
```

```python
#CALCULATING REPAIR COST PER SERVICE HOUR
df["repair_cost_per_hour"] = (
    df["repair_cost_inr"] /
    df["service_duration_hours"]
)

print("\nRepair Cost per Service Hour:")
print(
    df[
        ["service_id", "service_type",
         "repair_cost_inr",
         "service_duration_hours",
         "repair_cost_per_hour"]
    ]
)
```

```python
#FINAL DATASET
print("\nFinal Dataset:")
print(df)
```
